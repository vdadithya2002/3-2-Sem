#Text Processing will happen in 5 steps

###1.Tokenization
###2.Text Removal
###3.Stemming
###4.Lemmatization
###5.Pos Tagging

##Step 1 :- Selecting the Corpus

```python
corpus = [
    "I can't wait for the new season of my favorite show!",
    "The COVID-19 pandemic has affected millions of people worldwide.",
    "U.S. stocks fell on Friday after news of rising inflation.",
    "<html><body>Welcome to the website!</body></html>",
    "Python is a great programming language!!! ??"
]
```

##Step 2:- Text Cleaning

tasks:- convering the text to lowercase , remove punctuation , numbers , special characters and HTML tags

```python
import re
import string
from bs4 import BeautifulSoup

def clean_text(text):
  text = text.lower() #for changing into lowercase
  text = re.sub(r'\d+', '', text) #For removing the numbers
  text = text.translate(str.maketrans('','', string.punctuation)) # to remove the punctuation
  text = re.sub(r'\W', ' ', text) # to remove the special characters
  text = BeautifulSoup(text, "html.parser").get_text() #to remove the html tags

  return text
cleaned_corpus = [clean_text(doc) for doc in corpus]
print(cleaned_corpus)

```

##Step 3:- Tokenization

Tasks:- to split the cleaned text into tokens and tokenize the tokens into the cleaned corpus


```python
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')

tokenized_corpus = [word_tokenize(doc) for doc in cleaned_corpus]
print(tokenized_corpus)
```

##Step 4:- Stop Words Removal

tasks:- Remove the common stop words from the corpus

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
filtered_corpus = [[word for word in doc if word not in stop_words] for doc in tokenized_corpus]
print(filtered_corpus)
```

##Step 5:- Stemming and Lemmatization

Tasks:- To reduce words into their base form using the stemming and lemmatization

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet')

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_corpus = [[stemmer.stem(word) for word in doc] for doc in filtered_corpus]
lemmatized_corpus = [[lemmatizer.lemmatize(word) for word in doc] for doc in filtered_corpus]

print(stemmed_corpus)
print(lemmatized_corpus)
```

##Step 6:- Handling Contractions

Tasks:- Expanding contradctions in the text

```python
import sys
!{sys.executable} -m pip install contractions

import contractions

expanded_corpus = [contractions.fix(doc) for doc in cleaned_corpus]
print(expanded_corpus)
```

##Step 7:- Handling Emojis and Emoticons

Tasks:- Converting emojis to their textual representation

```python
import sys
!{sys.executable} -m pip install emoji
import emoji

emoji_corpus = [emoji.demojize(doc) for doc in cleaned_corpus]
print(emoji_corpus)
```

##Step 8:- Spell Checking


Tasks:- Correcting spelling erros in the text


```python
import sys
!{sys.executable} -m pip install pyspellchecker
from spellchecker import SpellChecker

spell = SpellChecker()
corrected_corpus = [[spell.correction(word) for word in doc] for doc in tokenized_corpus]
print(corrected_corpus)
```
