##Experiment 1

```python
import nltk
from nltk.util import ngrams
from collections import defaultdict, Counter

# Step 1: Create your own corpus
corpus = [
    "I love natural language processing",
    "I love machine learning",
    "natural language processing is interesting",
    "machine learning is powerful",
    "I enjoy learning new things"
]

# Step 2: Preprocess corpus
tokens = []
for sentence in corpus:
    tokens.extend(sentence.lower().split())

# Step 3: Generate trigrams
trigrams = list(ngrams(tokens, 3))

# Step 4: Build trigram model
model = defaultdict(Counter)

for w1, w2, w3 in trigrams:
    model[(w1, w2)][w3] += 1

# Step 5: Function to predict next word
def predict_next_word(w1, w2):
    if (w1, w2) in model:
        return model[(w1, w2)].most_common(1)[0][0]
    else:
        return "No prediction available"

# Step 6: Real-time prediction
print("Enter two words:")
w1 = input("Word 1: ").lower()
w2 = input("Word 2: ").lower()

prediction = predict_next_word(w1, w2)

print("Predicted next word:", prediction)
```

##Experiment 2

```python
import nltk
from nltk.util import ngrams
from collections import defaultdict, Counter

# Step 1: Create your own corpus
corpus = [
    "i want to eat ice cream",
    "i want to eat i scream",
    "i love natural language processing",
    "i love machine learning",
    "natural language processing is interesting",
    "machine learning is powerful",
    "i enjoy learning new things"
]

# Step 2: Preprocess corpus
tokens = []
for sentence in corpus:
    tokens.extend(sentence.lower().split())

# Step 3: Generate bigrams
bigrams = list(ngrams(tokens, 2))

# Step 4: Build bigram model
# The model will store word frequencies: {w1: {w2: count}}
model = defaultdict(Counter)

for w1, w2 in bigrams:
    model[w1][w2] += 1

# Step 5: Function to predict next word
def predict_next_word(w1):
    if w1 in model:
        # Return the most common word following w1
        return model[w1].most_common(1)[0][0]
    else:
        return "No prediction available"

# Step 6: Real-time prediction
print("Enter one word:")
w1 = input("Word 1: ").lower()

prediction = predict_next_word(w1)

print("Predicted next word:", prediction)
```

##Experiment 3

```python
corpus = {
    'cricket': [
        "Cricket is a bat-and-ball game played between two teams of eleven players on a field with a 22-yard pitch in the center.",
        "A cricket match is contested by two teams, each with eleven players.",
        "The objective of the game is for the batting side to score runs while the bowling and fielding side tries to prevent this.",
        "Key terms include 'wicket', 'over', 'bowler', 'batsman', 'umpire', and 'boundary'.",
        "Test cricket, One Day Internationals, and Twenty20 are the main formats."
    ],
    'football': [
        "Football, also known as soccer, is a team sport played with a spherical ball between two teams of 11 players.",
        "The game is played on a rectangular field called a pitch, with a goal at each end.",
        "The object of the game is to score by getting the ball into the opposing goal.",
        "Players cannot use their hands or arms to propel the ball, except for the goalkeepers within their penalty area.",
        "Major tournaments include the FIFA World Cup and the UEFA Champions League."
    ],
    'basketball': [
        "Basketball is a team sport in which two teams, most commonly of five players each, oppose each other on a rectangular court.",
        "The objective is to shoot a ball through a hoop 10 feet (3.05 m) high mounted to a backboard at each end.",
        "Points are scored by shooting the ball through the opponent's basket; a three-point field goal, a two-point field goal, or one-point free throws.",
        "The NBA is the premier professional basketball league in the world.",
        "Dribbling and passing are fundamental skills."
    ]
}

print("Corpus created with the following categories:", corpus.keys())
print("Number of documents in cricket category:", len(corpus['cricket']))
print("Number of documents in football category:", len(corpus['football']))
print("Number of documents in basketball category:", len(corpus['basketball']))
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download necessary NLTK data
# The try-except blocks prevent re-downloading if already present
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')
try:
    nltk.data.find('corpora/wordnet')
except LookupError:
    nltk.download('wordnet')
try:
    # Added to handle the punkt_tab error encountered previously
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')

print("NLTK data 'punkt', 'stopwords', 'wordnet', and 'punkt_tab' downloaded (if not already present).")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Initialize the WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

# Load English stop words into a set for efficient lookup
stop_words = set(stopwords.words('english'))

# Create empty lists to store preprocessed documents and their labels
preprocessed_documents = []
labels = []

# Iterate through each sport category and its documents in the corpus
for category, documents in corpus.items():
    for doc in documents:
        # a. Convert the document text to lowercase
        lower_cased_doc = doc.lower()

        # b. Tokenize the lowercase text
        tokens = word_tokenize(lower_cased_doc)

        # c. Remove any tokens that are present in the English stop words set
        filtered_tokens = [word for word in tokens if word.isalpha() and word not in stop_words]

        # d. Lemmatize each remaining token to its base form
        lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

        # e. Join the list of lemmatized tokens back into a single string
        processed_doc = ' '.join(lemmatized_tokens)

        # f. Append this processed string to the preprocessed_documents list
        preprocessed_documents.append(processed_doc)
        # and the current sport category to the labels list
        labels.append(category)

# Print the first few entries from preprocessed_documents and labels to inspect the results
print("First 5 preprocessed documents:")
for i in range(min(5, len(preprocessed_documents))):
    print(f"Document {i+1} ({labels[i]}): {preprocessed_documents[i]}")

print("\nTotal documents preprocessed:", len(preprocessed_documents))
print("Total labels:", len(labels))
```

#####TF-IDF Vectorization


```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Instantiate TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit the vectorizer to preprocessed_documents and transform the documents into a TF-IDF matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(preprocessed_documents)

# Print the shape of the tfidf_matrix to verify the dimensions
print("Shape of TF-IDF matrix:", tfidf_matrix.shape)
print("Number of documents:", tfidf_matrix.shape[0])
print("Number of unique terms (features):", tfidf_matrix.shape[1])
```

##### Train Classification Model

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB

# 1. Convert labels to numerical format
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(labels)

print("Original labels:", labels[:5])
print("Encoded labels:", encoded_labels[:5])
print("Label classes:", label_encoder.classes_)

# 2. Split the TF-IDF vectorized data and encoded labels into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(tfidf_matrix,
                                                    encoded_labels,
                                                    test_size=0.15, # Using 25% for testing
                                                    random_state=42, # For reproducibility
                                                    stratify=encoded_labels) # To maintain class distribution

print("\nShape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)

# 3. Choose and instantiate a classification model (e.g., Multinomial Naive Bayes)
model = MultinomialNB()

# 4. Train the model using the training data
model.fit(X_train, y_train)

print("\nMultinomial Naive Bayes model trained successfully.")
```

##### Model Evaluation


```python
from sklearn.metrics import accuracy_score, classification_report

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model performance
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=label_encoder.classes_)

print(f"\nModel Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(report)
```
